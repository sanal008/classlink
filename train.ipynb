```python
import cv2
import numpy as np
import tensorflow as tf

# ================= CONFIG =================
MODEL_PATH = "/content/drive/MyDrive/sanal/classlink_yolov8/weights/best_saved_model/best_float32.tflite"
IMAGE_PATH = "/content/classlink-1/train/images/abel_001_jpg.rf.4f6a26d6d65d7c6cc4fcd401b7238e71.jpg"

IMG_SIZE = 640
CONF_THRESHOLD = 0.25

CLASS_NAMES = ['Abel', 'Alen', 'Anamika', 'Athul']
# ==========================================


# Load TFLite model
interpreter = tf.lite.Interpreter(model_path=MODEL_PATH)
interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("Input shape:", input_details[0]['shape'])
print("Output shape:", output_details[0]['shape'])

# Load image
image = cv2.imread(IMAGE_PATH)
orig_h, orig_w, _ = image.shape

# Preprocess (NHWC, float32, 0–1)
img = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = img.astype(np.float32) / 255.0
img = np.expand_dims(img, axis=0)

# Run inference
interpreter.set_tensor(input_details[0]['index'], img)
interpreter.invoke()

# Get output
output = interpreter.get_tensor(output_details[0]['index'])[0]  # (8, 8400)
output = output.T  # (8400, 8)

# Debug: check max confidence
print("Max raw class score:", np.max(output[:, 4:]))

detections = []

for det in output:
    x, y, w, h = det[:4]
    class_scores = det[4:]   # YOLOv8: NO objectness

    class_id = np.argmax(class_scores)
    score = class_scores[class_id]

    if score < CONF_THRESHOLD:
        continue

    # Convert xywh (normalized) → xyxy (pixels)
    x1 = int((x - w / 2) * orig_w)
    y1 = int((y - h / 2) * orig_h)
    x2 = int((x + w / 2) * orig_w)
    y2 = int((y + h / 2) * orig_h)

    detections.append({
        "class_id": class_id,
        "class_name": CLASS_NAMES[class_id],
        "confidence": float(score),
        "box": [x1, y1, x2, y2]
    })

    cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(
        image,
        f"{CLASS_NAMES[class_id]} {score:.2f}",
        (x1, max(y1 - 10, 0)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2
    )

# Save result
cv2.imwrite("tf_tflite_output.jpg", image)

# Print detections
print("\nDetections:")
for d in detections:
    print(d)

print("\nSaved output image as tf_tflite_output.jpg")
```
